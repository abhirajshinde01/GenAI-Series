```python
from dotenv import load_dotenv
load_dotenv()  # take environment variables from .env.
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a: int, b: int):
    """
        it will return sum of 2 numbers

        Args: 
            a (int): first number
            b (int): second number
    """
    return a + b

@tool
def multiply_numbers(a: int, b: int):
    """
        it will return product of 2 numbers

        Args: 
            a (int): first number
            b (int): second number
    """
    return a * b
```

```python
add_numbers.invoke({"a": 5, "b": 3})  # Example usage of the tool
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite", streaming=True)
```

```python
agent= create_agent(
    model = llm, 
    tools=[add_numbers, multiply_numbers], 
    system_prompt="You are a Math teacher and always use tool for calculations."
    )
```

```python
response = agent.invoke({"messages": [{"role": "user", "content": "What is 5 + 30 * 2?"}]})
```
