##### Connect with Ollama model

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="phi4-mini", base_url="http://localhost:11434",
    temperature=0,
)

res = llm.invoke("What is the capital of France?")

print(res.content)
```

```python
res = llm.invoke("Who is the president of the United States?")
print(res.content)
```

##### Static Prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
prompts = [
    ("system","You are a python developer."),  ### "system","ai","user"
    ("user","how to identify even numbers?")
]

res = llm.invoke(prompts)
print(res.content)
```
