##### Prompt Template - Dynamic Prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
out = StrOutputParser()
```

```python
#### My self chain node

def transform_case(result: str):
    return result.upper()
```

```python
prompts = ChatPromptTemplate.from_messages([
       {   
           "role": "system",
           "content": "You are a translator and you translate in {language}."
       },
       {
           "role": "user",
           "content": "{query}"
       }
])
final_prompts = prompts.format_messages(language="marathi", query="I love programming.")
print(final_prompts)
res = llm.invoke(final_prompts)
print(res.content)
```

```python
res = llm.invoke(prompts.invoke({"language":"marathi", "query":"I love programming."}))
print (res.content)
```

```python
chain = prompts | llm | out | transform_case
res = chain.invoke({"language":"Hinglish", "query":"I love langchain."})
print (res)
```
