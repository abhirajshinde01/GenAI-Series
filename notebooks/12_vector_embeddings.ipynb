```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [
    "ML is subset of AI",
    "DL uses neural networks",
    "Python is a programming language",
    "India won the cricket world cup in 1983",
    "AI models can understand and generate human language"
]
```

```python
from langchain_ollama import OllamaEmbeddings
embeddings = OllamaEmbeddings(model="nomic-embed-text")
```

```python
vector = embeddings.embed_documents(documents)
len(vector[0])
```

#### Store vector embeddings in vector DB

```python
from langchain_chroma import Chroma
```

```python
vectorstore = Chroma.from_texts(
    texts=documents,
    embedding=embeddings,
    persist_directory="./vector_db"
)
```

```python
query ="DL"
result = vectorstore.similarity_search(query=query, k=2)
result
```
