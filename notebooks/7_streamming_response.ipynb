```python
from dotenv import load_dotenv
load_dotenv()  # take environment variables from .env.
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite", streaming=True)  # streaming=True to enable streaming responses
```

```python
question = "What is gen AI ?"
```

```python
res = llm.stream(question)
for chunky in res:
    print(chunky.text, end="")  # Print each chunk as it arrives")
```

```python
from langchain_groq import ChatGroq
groq_llm = ChatGroq(model="openai/gpt-oss-20b")
```

```python
question = "What is gen AI ?"
res = groq_llm.stream(question)
for chunky in res:
    print(chunky.text, end="") 
```
